You are interested in the relationship between payment type and the fare amount the customer pays. One approach is to look at the average fare amount for each payment type. 

```python
taxi_data.groupby('payment_type')['fare_amount'].mean()
```

```python
print(taxi_data.describe())
print(taxi_data.info())
print(taxi_data['payment_type'].value_counts())
```

Based on the averages shown, it appears that customers who pay in credit card tend to pay a larger fare amount than customers who pay in cash. However, this difference might arise from random sampling, rather than being a true difference in fare amount. To assess whether the difference is statistically significant, you conduct a hypothesis test.


### Task 3. Hypothesis testing

Before you conduct your hypothesis test, consider the following questions where applicable to complete your code response:

1. Recall the difference between the null hypothesis and the alternative hypotheses. Consider your hypotheses for this project as listed below.

$H_0$: There is no difference in the average fare amount between customers who use credit cards and customers who use cash.

$H_A$: There is a difference in the average fare amount between customers who use credit cards and customers who use cash.



Your goal in this step is to conduct a two-sample t-test. Recall the steps for conducting a hypothesis test: 


1.   State the null hypothesis and the alternative hypothesis
2.   Choose a signficance level
3.   Find the p-value
4.   Reject or fail to reject the null hypothesis 



**Note:** For the purpose of this exercise, your hypothesis test is the main component of your A/B test. 

You choose 5% as the significance level and proceed with a two-sample t-test.

```python
credit_card = taxi_data[taxi_data['payment_type']==1]['fare_amount']
cash = taxi_data[taxi_data['payment_type']==2]['fare_amount']
stats.ttest_ind(a = credit_card, b = cash,alternative = 'two-sided', equal_var = False)
```

The difference in the average fare amount between customers who use credit cards and those who use cash is statistically significant. The data shows that credit card users tend to pay higher fares on average, completely ruling out random sampling luck.

WHY A/B TESTING MIGHT NOT BE REALISTIC

While this layout works perfectly for an educational exercise, it breaks several core principles of realistic world A/B testing.
 In a real A/B test, the experimenter must randomly assign users to an experience (e.g. hiding cash as an option for half the users). In this dataset, the data professional simply observed a customer's pre-existing natural preference. This makes it an observational study, not a true controlled experiment.
The test assumes the payment method causes the different fare amounts. In reality, other hidden factors are likely at play. For example, a customer taking an expensive $100 ride to the airport is naturally going to use a credit card because carrying that much cash is inconvenient. Conversely, short, cheap $5 cross town trips are much more likely to be settled with cash.
Many ride hailing datasets automatically include electronic tips in the credit card fare_amount tracking columns, whereas cash tips are rarely recorded by drivers in the system. This data entry bias can make cards appear artificially higher than they actually are.
